### 4.1. Número de libros publicados:

El primer análisis determinará cuántos libros se publicaron después del 1 de enero de 2000. Este resultado permitirá conocer el volumen de publicaciones recientes dentro de la base de datos.

```python
# Contar los libros publicados después del 1 de enero de 2000
query_1 = """
SELECT COUNT(*) AS total_libros
FROM books
WHERE publication_date > DATE '2000-01-01';
"""

result_1 = pd.read_sql_query(query_1, engine)
display(result_1)
```

Se realizó una consulta SQL para calcular la cantidad de libros publicados después del 1 de enero del año 2000. Para ello, se utilizó la tabla books y se aplicaron las siguientes instrucciones:
- COUNT(*): cuenta el número de registros que cumplen con la condición establecida.
- AS total_libros: asigna el nombre total_libros a la columna que contiene el resultado del conteo.
- FROM books: indica que la información se obtiene de la tabla books.
- WHERE publication_date > DATE '2000-01-01': filtra los registros para incluir únicamente los libros cuya fecha de publicación sea posterior al 1 de enero de 2000.
Posteriormente, se utilizó pd.read_sql_query(query_1, engine) para ejecutar la consulta en PostgreSQL y almacenar el resultado en el DataFrame result_1. Finalmente, mediante display(result_1), se mostró la cantidad total de libros que cumplen con el criterio establecido.

Con esta consulta se obtuvo un indicador del volumen de libros publicados después de la fecha de referencia, lo que permite conocer parte de la composición temporal del catálogo disponible en la base de datos.

### 4.2. Número de reseñas y calificación promedio:

En este apartado se analizará la recepción de cada libro por parte de los usuarios. Se calculará el número de reseñas escritas y la calificación promedio recibida por cada título.

Para evitar duplicaciones, las reseñas y las calificaciones se agregarán por separado antes de combinar sus resultados.

```python
# Contar el número total de reseñas de texto
query_reviews = """
SELECT COUNT(*) AS total_resenas
FROM reviews;
"""

result_reviews = pd.read_sql_query(query_reviews, engine)
display(result_reviews)
```

```python
# Obtener el número de reseñas y la calificación promedio por libro
query_2 = """
WITH review_counts AS (
    SELECT
        book_id,
        COUNT(*) AS total_resenas
    FROM reviews
    GROUP BY book_id
),
rating_averages AS (
    SELECT
        book_id,
        AVG(rating) AS calificacion_promedio
    FROM ratings
    GROUP BY book_id
)
SELECT
    b.book_id,
    b.title,
    COALESCE(rc.total_resenas, 0) AS total_resenas,
    ROUND(ra.calificacion_promedio::numeric, 2)
        AS calificacion_promedio
FROM books AS b
LEFT JOIN review_counts AS rc
    ON b.book_id = rc.book_id
LEFT JOIN rating_averages AS ra
    ON b.book_id = ra.book_id
ORDER BY total_resenas DESC, b.title;
"""

result_2 = pd.read_sql_query(query_2, engine)
display(result_2)
```

Se realizó una consulta SQL para obtener el número de reseñas y la calificación promedio de cada libro registrado en la base de datos. El objetivo fue conocer la cantidad de opiniones que ha recibido cada libro y evaluar su valoración promedio por parte de los usuarios.

Para desarrollar la consulta, se utilizaron las siguientes instrucciones:
- WITH review_counts AS (...): crea una expresión de tabla común (CTE) que agrupa las reseñas por book_id y utiliza COUNT(*) para calcular el número total de reseñas de cada libro.
- WITH rating_averages AS (...): calcula la calificación promedio de cada libro mediante AVG(rating), agrupando los registros por book_id.
- SELECT: selecciona el identificador y el título de cada libro, además del total de reseñas y la calificación promedio.
- FROM books AS b: establece la tabla books como base de la consulta y le asigna el alias b para facilitar su referencia.
- LEFT JOIN: combina la información de los libros con los resultados de las reseñas y las calificaciones, conservando todos los libros de la tabla books, incluso si no tienen reseñas o calificaciones registradas.
- COALESCE(rc.total_resenas, 0): sustituye por cero los valores nulos del total de reseñas cuando un libro no tiene ninguna registrada.
- ROUND(ra.calificacion_promedio::numeric, 2): redondea la calificación promedio a dos decimales para facilitar su interpretación.
- ORDER BY total_resenas DESC, b.title: ordena los resultados de mayor a menor según el número de reseñas y, en caso de empate, alfabéticamente por título.
Posteriormente, se utilizó pd.read_sql_query(query_2, engine) para ejecutar la consulta en PostgreSQL y almacenar los resultados en el DataFrame result_2. Finalmente, mediante display(result_2), se presentó la información en una tabla dentro del notebook.

Con este análisis se obtuvo una visión general de la participación de los usuarios y de las valoraciones de los libros. Los resultados permiten identificar cuáles cuentan con más reseñas y comparar sus calificaciones promedio, lo que contribuye a comprender mejor la interacción de los lectores con los libros del catálogo.

### 4.3. Editorial que publicó más libros de más de 50 páginas:

Se identificará la editorial con el mayor número de libros que superan las 50 páginas. Este filtro permite excluir folletos y publicaciones demasiado breves para concentrar el análisis en libros de mayor extensión.

```python
# Identificar la editorial con más libros de más de 50 páginas
query_3 = """
SELECT
    p.publisher,
    COUNT(b.book_id) AS total_libros
FROM publishers AS p
INNER JOIN books AS b
    ON p.publisher_id = b.publisher_id
WHERE b.num_pages > 50
GROUP BY p.publisher_id, p.publisher
ORDER BY total_libros DESC
LIMIT 1;
"""

result_3 = pd.read_sql_query(query_3, engine)
display(result_3)
```

Se realizó una consulta SQL para identificar la editorial que ha publicado la mayor cantidad de libros con más de 50 páginas. El objetivo fue conocer qué editorial tiene la mayor presencia dentro del catálogo, considerando únicamente los libros que cumplen con este criterio de extensión.

Para desarrollar la consulta, se utilizaron las siguientes instrucciones:
- SELECT p.publisher: selecciona el nombre de la editorial para mostrarlo en los resultados.
- COUNT(b.book_id) AS total_libros: cuenta los libros que cumplen con la condición establecida y asigna el nombre total_libros a la columna resultante.
- FROM publishers AS p: establece la tabla publishers como punto de partida y utiliza el alias p para simplificar su referencia.
- INNER JOIN books AS b ON p.publisher_id = b.publisher_id: combina las tablas publishers y books mediante el identificador de la editorial, considerando únicamente los registros que tienen correspondencia en ambas tablas.
- WHERE b.num_pages > 50: filtra los libros para incluir solamente aquellos que tienen más de 50 páginas.
- GROUP BY p.publisher_id, p.publisher: agrupa los registros por editorial para calcular el total de libros publicados por cada una.
- ORDER BY total_libros DESC: ordena las editoriales de mayor a menor según la cantidad de libros que cumplen con el criterio.
- LIMIT 1: devuelve únicamente la editorial que ocupa el primer lugar en el ordenamiento.

Posteriormente, se utilizó pd.read_sql_query(query_3, engine) para ejecutar la consulta en PostgreSQL y almacenar el resultado en el DataFrame result_3. Finalmente, mediante display(result_3), se visualizó la editorial identificada y el número de libros correspondiente.

Con este análisis se determinó cuál es la editorial con más libros de más de 50 páginas dentro de la base de datos. Este resultado permite conocer qué editorial tiene la mayor cantidad de publicaciones que cumplen con el criterio establecido y contribuye a comprender la distribución del catálogo editorial.

### 4.4. Autor con la calificación promedio más alta:

En este análisis se identificará al autor cuyos libros tienen la calificación promedio más alta, considerando exclusivamente los libros que recibieron al menos 50 calificaciones.

El filtro permitirá concentrarse en libros con una cantidad mínima de valoraciones, reduciendo la influencia de resultados basados en muy pocas opiniones.

```python
# Identificar al autor con la calificación promedio más alta,
# considerando libros con al menos 50 calificaciones

query_4 = """
WITH book_ratings AS (
    SELECT
        book_id,
        AVG(rating) AS calificacion_promedio,
        COUNT(rating_id) AS total_calificaciones
    FROM ratings
    GROUP BY book_id
    HAVING COUNT(rating_id) >= 50
),
author_ratings AS (
    SELECT
        a.author_id,
        a.author,
        AVG(br.calificacion_promedio) AS calificacion_promedio,
        COUNT(br.book_id) AS libros_evaluados,
        SUM(br.total_calificaciones) AS total_calificaciones
    FROM authors AS a
    INNER JOIN books AS b
        ON a.author_id = b.author_id
    INNER JOIN book_ratings AS br
        ON b.book_id = br.book_id
    GROUP BY a.author_id, a.author
)
SELECT
    author,
    ROUND(calificacion_promedio::numeric, 2)
        AS calificacion_promedio,
    libros_evaluados,
    total_calificaciones
FROM author_ratings
ORDER BY calificacion_promedio DESC
LIMIT 1;
"""

result_4 = pd.read_sql_query(query_4, engine)
display(result_4)
```

El análisis identificó a J.K. Rowling/Mary GrandPré como el autor con la calificación promedio más alta, con un resultado de 4.28, considerando cuatro libros que recibieron al menos 50 calificaciones cada uno.

El número de libros evaluados proporciona contexto adicional para interpretar este resultado, ya que permite conocer cuántas publicaciones respaldan la calificación promedio obtenida. Aunque el resultado refleja una valoración favorable de los libros considerados, es importante tener en cuenta que se basa en cuatro títulos elegibles y no necesariamente representa toda la obra del autor.

Este indicador permite complementar la comparación entre autores y realizar una interpretación más informada de las calificaciones de los lectores.

### 4.5. Promedio de reseñas de texto entre usuarios que calificaron más de 50 libros: 

Por último, se analizará la participación de los usuarios más activos en cuanto a calificaciones. Se identificarán los usuarios que calificaron más de 50 libros distintos y se calculará el promedio de reseñas de texto que realizaron.

Para que el promedio represente a todos los usuarios que cumplen el criterio, también se incluirán aquellos que no hayan escrito reseñas de texto.

```python
# Calcular el promedio de reseñas de texto de usuarios
# que calificaron más de 50 libros distintos
query_5 = """
WITH active_users AS (
    SELECT
        username
    FROM ratings
    GROUP BY username
    HAVING COUNT(DISTINCT book_id) > 50
),
user_reviews AS (
    SELECT
        au.username,
        COUNT(DISTINCT r.review_id) AS total_resenas
    FROM active_users AS au
    LEFT JOIN reviews AS r
        ON au.username = r.username
    GROUP BY au.username
)
SELECT
    COUNT(*) AS usuarios_analizados,
    ROUND(AVG(total_resenas), 2) AS promedio_resenas_texto
FROM user_reviews;
"""

result_5 = pd.read_sql_query(query_5, engine)
display(result_5)
```

Se realizó una consulta SQL para calcular el promedio de reseñas de texto escritas por los usuarios que han calificado más de 50 libros distintos. El objetivo fue analizar la participación de los usuarios más activos en cuanto a la publicación de reseñas escritas dentro de la plataforma.

Para desarrollar la consulta, se utilizaron las siguientes instrucciones:
- WITH active_users AS (...): crea una expresión de tabla común (CTE) para identificar a los usuarios que han calificado más de 50 libros diferentes.
- GROUP BY username: agrupa los registros de la tabla ratings por nombre de usuario.
- HAVING COUNT(DISTINCT book_id) > 50: selecciona únicamente a los usuarios que han calificado más de 50 libros distintos, evitando contar varias veces un mismo libro.
- user_reviews AS (...): crea una segunda CTE para calcular el número de reseñas escritas por cada usuario seleccionado.
- LEFT JOIN reviews AS r ON au.username = r.username: relaciona a los usuarios activos con la tabla reviews mediante el nombre de usuario, conservando también a quienes no han escrito reseñas.
- COUNT(DISTINCT r.review_id) AS total_resenas: cuenta las reseñas de texto únicas de cada usuario.
- GROUP BY au.username: agrupa los resultados por usuario para obtener su total individual de reseñas.
- COUNT(*) AS usuarios_analizados: calcula cuántos usuarios cumplen con el criterio de haber calificado más de 50 libros distintos.
- AVG(total_resenas): calcula el promedio de reseñas escritas por los usuarios identificados.
- ROUND(AVG(total_resenas), 2): redondea el promedio a dos decimales para facilitar su interpretación.

Posteriormente, se utilizó pd.read_sql_query(query_5, engine) para ejecutar la consulta en PostgreSQL y almacenar los resultados en el DataFrame result_5. Finalmente, mediante display(result_5), se visualizaron el número de usuarios analizados y el promedio de reseñas de texto por usuario.

Con este análisis se obtuvo un indicador de la participación escrita de los usuarios que califican más de 50 libros distintos. Los resultados permiten conocer cuántos usuarios cumplen con este criterio y cuántas reseñas escritas publican, en promedio, lo que ayuda a comprender mejor sus hábitos de interacción con la plataforma.
